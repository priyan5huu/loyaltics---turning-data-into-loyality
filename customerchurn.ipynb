```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
#loading the datset

df = pd.read_csv("telco.csv")
df.head() #first 5 rows
```

```python
#shape of dataset

df.shape
```

```python
# finding null values

df.isnull().sum()
```

```python
# finding duplicate values

df.duplicated().sum()
```

```python
df = df.drop("customerID", axis=1) # as customer id has no major role so we drop it
df.head()
```

```python
df.dtypes
```

```python
#converting to number
df.TotalCharges.values
```

```python
#coerce to ignore the erros as

pd.to_numeric(df.TotalCharges, errors="coerce").isnull().sum()
```

```python
df = df[df.TotalCharges!=' ']
df.shape
```

```python
df.TotalCharges = pd.to_numeric(df.TotalCharges)
df.dtypes
```

```python
#value count

df.Churn.value_counts()
```

```python
loyal_customer = df[df.Churn=='No'].tenure  #checking tenure of loyal customer
churned_customer = df[df.Churn=='Yes'].tenure #checking tenure of churned
plt.xlabel('Tenure')
plt.ylabel('No of customers')
plt.title('Tenure vs No of customers')
plt.hist([loyal_customer, churned_customer], color=['green', 'red'], label=['loyal', 'churned'])
plt.legend()
```

```python
#label encoding
#we used this for loop to know
for column in df :
  print(f'{column} : {df[column].unique()}')
```

```python
df.replace('No internet service', 'No', inplace=True)
df.replace('No phone service', 'No', inplace=True)
```

```python
for column in df :
  print(f'{column} : {df[column].unique()}')
```

```python
yes_no_columns = ['Partner', 'Dependents', 'PhoneService', 'MultipleLines',
                  'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport',
                  'StreamingTV', 'StreamingMovies', 'PaperlessBilling', 'Churn']

for column in yes_no_columns:
  df[column].replace({'Yes': 1, 'No': 0}, inplace=True)
```

```python
for column in df :
  print(f'{column} : {df[column].unique()}')
```

```python
df['gender'].replace({'Female': 1, 'Male': 0}, inplace=True)
df.gender.unique()
```

```python
df = pd.get_dummies(df, columns=['InternetService', 'Contract', 'PaymentMethod'])
```

```python
#scaling of features

cols_to_scale = ['tenure', 'MonthlyCharges', 'TotalCharges']

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df[cols_to_scale] = scaler.fit_transform(df[cols_to_scale])
```

```python
X = df.drop('Churn', axis = 1)
y = df['Churn']
print(f'Shape of X after dropping Churn: {X.shape}')
```

```python
#train test split - kitna data training me jaiga and kitno data testing me jaigaa

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=5)
print(f'Shape of X_train: {X_train.shape}')
print(f'Shape of X_test: {X_test.shape}')
print('\nDtypes of X_train after split:')
print(X_train.dtypes)
```

```python
print(y_train.dtype)
print(y_train.unique())
```

```python
import tensorflow as tf
from tensorflow import keras

model = keras.Sequential([
    keras.layers.Dense(26, input_shape=(26,), activation='relu'),
    keras.layers.Dense(15, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

model.fit(X_train, y_train, epochs=100)
```
